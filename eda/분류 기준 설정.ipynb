## 중분류별 소분류 매출 비중 EDA

이 절에서는 `target_product_0520_merged.csv`를 기준으로, 하나의 중분류 안에서 소분류가 `출시후_30일_매출액_합계`의 몇 퍼센트를 차지하는지 확인했다. 현재 파일은 상품명 중복을 병합한 뒤의 데이터이며, 분석 대상은 2,304개 상품, 중분류는 46개다.

집계는 `상품중분류명 + 상품소분류명` 단위로 먼저 매출액을 합산한 뒤, 각 중분류 내부 총매출을 100%로 두고 소분류별 비중을 계산했다. 이때 `출시후_30일_매출액_합계`가 실제로 0인 소분류 조합은 그래프와 순위 계산에서 제외했다. 제외된 조합은 `냉장주스-유산균음료`이다.

색상은 실제 소분류명이 아니라 각 중분류 내부의 비중 순위에 맞춰 통일했다. 즉 모든 막대에서 1위 색상은 해당 중분류 안에서 가장 매출 비중이 큰 소분류를 의미하지만, 실제 소분류명은 중분류마다 다를 수 있다.

이번 그래프는 단순 매출 규모순이 아니라 소분류 누적 분포가 비슷한 중분류끼리 묶었다. 묶음 기준은 1위 소분류 비중, 상위 3개 소분류 누적 비중, 소분류 수, 유효 소분류 수를 함께 사용했다. 독점형에서 분산형으로 갈수록 한 소분류에 매출이 몰리는 정도가 낮아진다.

## 매출 비중과 상품 수 비중 비교 EDA

앞의 두 EDA는 같은 중분류-소분류 구조를 보지만 기준값이 다르다. 첫 번째는 `출시후_30일_매출액_합계` 기준이고, 두 번째는 `상품명` 개수 기준이다. 이번 절에서는 두 기준이 중분류별로 얼마나 비슷하게 나타나는지, 그리고 어디서 달라지는지를 비교했다.

비교 지표는 중분류별 `1위 소분류 비중`, `상위 3개 소분류 누적 비중`, `1위 소분류명 일치 여부`, `분포 유형 전환`이다. 1위 소분류 비중의 상관계수는 0.81, 상위3 누적 비중의 상관계수는 0.79로 나타났다. 46개 중분류 중 매출 기준 1위 소분류와 상품 수 기준 1위 소분류가 같은 경우는 37개, 다른 경우는 9개다. 1위 비중 차이의 평균 절대값은 12.0%p, 중앙값은 9.8%p다.

## 소분류 기준점 분리 기준 검토

앞선 EDA의 목적은 “중분류를 그대로 [기준점]으로 둘 것인가, 아니면 일부 중분류는 소분류를 [기준점]으로 내려서 볼 것인가”를 정하는 것이다. 이 판단은 상품 수가 아니라 `출시후_30일_매출액_합계`의 소분류별 분포를 기준으로 한다. 성공/비성공 라벨 자체가 각 [기준점] 안에서 누적 매출 80%를 견인하는 상품을 성공으로 정의하기 때문에, 기준점의 적절성도 매출 분포의 동질성으로 판단하는 것이 논리적으로 일관된다.

여기서 기준점 분리의 이론적 핵심은 “분류 단위 내부의 이질성”이다. 하나의 중분류 안에서 사실상 한두 소분류가 대부분의 매출을 설명한다면, 그 중분류는 비교적 단일한 시장처럼 볼 수 있다. 반대로 최소 3개 이상의 소분류가 의미 있는 매출 비중을 나누어 갖는다면, 하나의 중분류 안에 여러 하위 시장이 공존한다고 볼 수 있다. 이 경우 중분류 단위로 파레토 80%를 적용하면 서로 성격이 다른 상품들이 같은 경쟁장에 놓이므로, 소분류 단위로 [기준점]을 분리하는 것이 더 타당하다.

이번 검토에서는 두 기준을 비교했다. 첫 번째 기준은 `유효 소분류 수 >= 3`이다. 이는 정보 엔트로피 기반 지표로, 소분류 개수가 아니라 매출 비중상 실제로 몇 개 소분류가 의미 있게 존재하는지를 본다. 두 번째 기준은 `매출 비중 15% 이상 소분류 수 >= 3`이다. 이는 전체 매출의 15% 이상을 차지하는 소분류가 최소 3개 이상인지 보는 기준으로, 발표와 보고서에서 설명하기 쉽다는 장점이 있다.

두 기준은 각각 15개, 15개 중분류를 후보로 선택했다. 두 기준이 공통으로 선택한 후보는 10개이고, 한 기준만 만족한 후보는 10개다.

### 기준 검토 그래프 1. 두 기준별 후보 통과 여부

![소분류 기준점 분리 후보 통과 여부](eda_outputs/split_criteria_candidate_map.png)

이 그래프는 산점도 대신 후보 중분류 20개를 행 단위로 놓고, 두 기준의 값을 나란히 비교한 것이다. 왼쪽은 `유효 소분류 수`, 오른쪽은 `15% 이상 소분류 수`이며, 각 패널의 점선은 기준 통과선인 3을 의미한다. 두 막대가 모두 점선을 넘으면 공통 후보이고, 한쪽만 넘으면 해당 기준으로만 도출된 후보이다. 이 방식은 어떤 중분류가 어떤 기준 때문에 선택되었는지 가장 직접적으로 보여준다.

### 기준 검토 그래프 2. 후보군 구성

![두 기준 후보군 구성](eda_outputs/split_criteria_group_counts.png)

공통 후보는 두 기준이 동시에 “소분류로 나누어 볼 만큼 내부 이질성이 크다”고 판단한 중분류다. 반면 기준 1만 만족하는 중분류는 전체 분포가 완만하게 퍼져 있지만 15% 이상으로 명확히 큰 소분류는 2개 이하인 경우다. 기준 2만 만족하는 중분류는 15% 이상 소분류가 3개 이상 존재하지만, 한두 소분류의 비중이 상대적으로 커서 유효 소분류 수는 3에 살짝 못 미치는 경우다.

### 기준 검토 그래프 3. 공통 후보 10개의 소분류 구성

![공통 후보 소분류 구성](eda_outputs/split_criteria_common_stacked.png)

공통 후보는 `가공미반류, 가공우유, 간식빵, 냉장간식, 냉장간편식, 냉장베이커리, 스낵류, 즉석치킨, 차음료, 초콜릿`이다. 이들은 두 기준이 모두 선택한 후보이므로, 소분류 기준점 분리의 우선순위가 가장 높다. 그래프에서 볼 수 있듯이 이 중분류들은 1위 소분류 하나가 압도적으로 지배하기보다는 2위, 3위, 때로는 4위 소분류까지 의미 있는 매출 비중을 갖는다. 따라서 중분류 하나로 파레토 80%를 적용하면 서로 다른 소분류의 상품들이 하나의 성공/비성공 경쟁장에 섞일 가능성이 크다.

### 기준 검토 그래프 4. 공통이 아닌 10개 후보의 소분류 구성

![공통이 아닌 후보 소분류 구성](eda_outputs/split_criteria_noncommon_stacked.png)

공통이 아닌 후보는 그래프 안에서 왼쪽의 `기준 1만 만족` 그룹과 오른쪽의 `기준 2만 만족` 그룹으로 분리해 표시했다. 기준 1만 만족한 `기능성드링크, 비스킷류, 용기면, 캔, 커피음료`와 기준 2만 만족한 `냉동간편식, 냉장안주, 냉장피자, 상온간편식, 젤리`이다. 기준 1만 만족한 그룹은 매출 분포가 꼬리까지 넓게 퍼져 있어서 유효 소분류 수가 높지만, 15% 이상을 넘는 뚜렷한 대형 소분류는 2개 정도인 구조다. 기준 2만 만족한 그룹은 보통 3개 소분류가 대부분의 매출을 나눠 갖지만, 그중 1위 또는 2위 비중이 상대적으로 커서 유효 소분류 수가 3에 미달한다.

### 해석 및 기준 선택

보수적으로 접근한다면 공통 후보 10개를 우선적으로 소분류 [기준점]으로 분리하는 것이 가장 안정적이다. 이 방식은 두 기준이 동시에 동의한 경우만 분리하므로 자의적 분리를 줄일 수 있다. 반면 더 적극적으로 기준점을 세분화하고 싶다면 `15% 이상 소분류 수 >= 3`을 주 기준으로 삼는 것이 설명력이 좋다. “전체 매출의 15% 이상을 담당하는 소분류가 3개 이상이면, 하나의 중분류로 묶기보다 소분류별 기준점으로 보는 것이 타당하다”는 문장으로 바로 설명할 수 있기 때문이다.

최종적으로는 공통 후보 10개를 1차 분리 대상으로 두고, 공통이 아닌 10개는 분석 목적에 따라 검토 후보로 두는 방식을 추천한다. 이렇게 하면 이론적으로는 내부 이질성이 충분히 큰 중분류를 우선 분리하고, 실무적으로는 너무 많은 기준점이 생기는 문제를 피할 수 있다.

### 해석

`유효 소분류 수 >= 3`은 매출 분포 관점에서 중분류 내부가 충분히 분산되어 있는지를 보는 기준이다. 다만 이것만으로는 소분류별 표본 크기까지 보장하지 않는다. 예를 들어 `냉장간편식`, `즉석치킨`은 매출 분포상 기준 1을 만족하지만, 일부 소분류의 상품 수가 1개에 불과하다. 이런 경우 해당 소분류를 독립된 [기준점]으로 만들면, 상품 1개만으로 성공/비성공 판단이 결정되어 모델링이나 평가 기준이 불안정해질 수 있다.

따라서 실무적으로는 기준 1을 그대로 적용하기보다, 소분류별 최소 상품 수 조건을 함께 두는 것이 안전하다. 예를 들어 `유효 소분류 수 >= 3`을 만족하더라도, 모든 소분류가 최소 3개 또는 5개 이상의 상품을 갖는지 확인한 뒤 분리 여부를 확정하는 방식이다. 보수적으로는 최소 상품 수 5개 이상을 권장할 수 있고, 데이터 수가 부족한 분석에서는 최소 3개 이상을 하한선으로 둘 수 있다.



## 1위 소분류 독점형 예외 처리 기준

앞서 설정한 소분류 [기준점] 분리의 기본 논리는 중분류 내부의 다양성 또는 이질성이다. 즉, 여러 소분류가 의미 있는 매출 비중을 나누어 가질 때 중분류 하나를 기준점으로 삼으면 서로 다른 하위 시장의 상품들이 같은 경쟁장에 놓이게 된다. 이를 해결하기 위해 `유효 소분류 수 >= 3` 또는 `15% 이상 소분류 수 >= 3`과 같은 분산형 기준을 검토했다.

하지만 이와 별도로 고려해야 할 예외가 있다. 특정 중분류에서 1위 소분류가 매출을 압도적으로 지배하는 경우다. 이 경우 중분류 내부가 다양하게 분산되어 있지는 않기 때문에 앞의 기준 1, 기준 2에는 걸리지 않을 수 있다. 그러나 중분류 전체를 기준으로 파레토 80% 성공 상품을 선정하면, 1위 소분류가 성공 영역 대부분을 차지하고 나머지 소분류의 상품은 구조적으로 비성공으로 밀릴 가능성이 있다.

이 문제는 “중분류 내부가 다양해서 나누는 문제”가 아니라 “지배적 소분류가 성공 판정 공간을 잠식하는 문제”에 가깝다. 따라서 기본 분리 기준과는 별도의 예외 처리 기준으로 정의하는 것이 타당하다.

### 예외 처리 기준

본 분석에서는 복잡한 사후 라벨 검증 기준 대신, 사전에 매출 구조만으로 판단 가능한 단순 기준을 사용한다.

다음 조건을 모두 만족하는 중분류는 1위 소분류 독점형 예외 분리 후보로 본다.

1. `1위 소분류 매출 비중 >= 80%`
2. `2위 이하 소분류 전체 매출 비중 >= 10%`
3. `2위 이하 소분류 전체 상품 수 >= 5개`

첫 번째 조건은 1위 소분류가 중분류 전체 매출을 사실상 지배하는지를 확인한다. 두 번째 조건은 나머지 소분류가 완전히 무시할 수준은 아닌지를 확인한다. 세 번째 조건은 나머지 소분류에도 최소한의 상품 수가 있어 별도 기준점으로 평가할 실익이 있는지를 확인한다.

### 처리 방식

이 예외 기준을 만족하는 경우, 모든 소분류를 개별 기준점으로 쪼개기보다는 다음과 같이 단순하게 나누는 방식을 우선 검토한다.

- 기준점 1: 1위 소분류
- 기준점 2: 비1위 소분류 묶음

이 방식은 기준점이 과도하게 늘어나는 문제를 줄이면서도, 1위 소분류의 압도적인 매출 지배 때문에 나머지 소분류 상품이 전부 비성공으로 밀리는 문제를 완화할 수 있다.

### 기본 분리 기준과의 관계

정리하면, 소분류 [기준점] 분리는 두 가지 논리로 나뉜다.

첫째, 분산형 분리 기준이다. 이는 중분류 안에 의미 있는 소분류가 3개 이상 존재하는 경우로, 중분류 내부 이질성이 높기 때문에 소분류 단위로 기준점을 나누는 방식이다.

둘째, 독점형 예외 기준이다. 이는 1위 소분류가 중분류 매출을 압도적으로 지배하지만, 나머지 소분류도 일정 규모 이상 존재하는 경우다. 이때는 다양성 때문이 아니라 1위 소분류의 성공 라벨 잠식을 방지하기 위해 예외적으로 `1위 소분류`와 `비1위 소분류 묶음`으로 나눈다.

따라서 최종 기준점 설정은 다음 순서로 적용할 수 있다.

1. 먼저 분산형 기준으로 소분류 분리 후보를 선정한다.
2. 분산형 기준에 걸리지 않은 중분류 중에서 1위 소분류 독점형 예외 기준을 검토한다.
3. 예외 기준을 만족하면 `1위 소분류`와 `비1위 소분류 묶음`으로 분리한다.
4. 둘 다 해당하지 않으면 중분류를 그대로 [기준점]으로 유지한다.

이렇게 하면 기준점 분리의 기본 목적, 즉 비교 가능한 상품끼리 성공/비성공을 판단한다는 원칙을 유지하면서도, 독점 소분류로 인한 구조적 불리함을 일정 부분 보정할 수 있다.
